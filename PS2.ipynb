Before beginning analysis, explain:

- What economic relationships might we expect to find?

## What is the unit of observation in this dataset?

The unit of observation is a particular airline route between two cities in a particular year

## Why might we use logs of variables instead of levels?

I can think of two reasons: one is that since the natural log of a negative number isn't defined, it makes sense to use log when we know that the variable is >0 which is the case for quantities, prices, etc. The other reason is that you can take expressions like a*x^2*y*2 and linearize it applying log --> log(a)+2*log(x)+2*log(y).  

## What economic relationships might we expect to find?

We can expect to see an increase in the number of passengers traveling by airplanes with greater economic growth and if people have enough disposable income to travel more. Studying the fare and number of passengers can also tell us if demand is elastic or inelastic. 

## What role might market concentration play in pricing?

It stands to reason that if an airline has a bigger market share, they might raise fares. The airline business is notoriously low margin, so having a monopoloy might result in airlines reducing the number of amenities they offer as they aren't worried about customers fleeing to a competitor.
